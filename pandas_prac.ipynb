Both `loc` and `iloc` are row-first, column-second. This is the opposite of what we do in native Python, which is column-first, row-second.

```python
medical_data.iloc[:,2]
```

```python
medical_data.iloc[2:5,2]
```

```python
medical_data.iloc[[1,3,7],2]
```

```python
medical_data.iloc[-5:]
```

```python
medical_data.loc[0,"Region"]
```

In `iloc` we use indices of rows/columns, but in `loc` we use the names of the rows/columns

```python
medical_data.loc[:,["Actual_Sales","ERP_Stat_FC","Bias_Adjusted_Stat_FC"]]
```

In this case `df.iloc[0:1000]` will return 1000 entries, while `df.loc[0:1000]` return 1001 of them! To get 1000 elements using loc, you will need to go one lower and ask for `df.loc[0:999]`
