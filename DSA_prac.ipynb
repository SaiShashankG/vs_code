````python
def mirror(n):
    s = str(n)
    mirror_str = s[::-1]
    return int(mirror_str)

def find_min_mirror_distance(nums): 
    min_distance=nums[0]
    for i, num1 in enumerate(nums):
        mirrored = mirror(num1)
    
        for j in range(i + 1, len(nums)):
            if nums[j] == mirrored:
                distance = j - i
                min_distance = min(min_distance, distance)
    
    return min_distance if min_distance != float('inf') else -1

nums=[9,9]
print(find_min_mirror_distance(nums))
````

````python
def bubble_sort(arr):
    n=len(arr)
    for i in range(n):
        isSwap=False
        for j in range(n-i-1):
            if arr[j]>arr[j+1]:
                isSwap=True
                arr[j],arr[j+1]=arr[j+1],arr[j]
        if not isSwap:
            break
    return arr
````

````python
def insertion_sort(arr):
    n=len(arr)
    for i in range(1,n):
        key=arr[i]
        j=i-1
        # Alternate Approach
        # while j>=0 and arr[j]>key:
        #     arr[j],arr[j+1]=arr[j+1],arr[j]
        #     j-=1
        while j>=0 and arr[j]>key:
            arr[j+1]=arr[j]
            j-=1
        arr[j+1]=key
    return arr
````

144. Given the root of a binary tree, return the preorder traversal of its nodes' values `LeetCode`

````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def __init__(self):
        self.ans=[]
    def preorder(self,root):
        #base case
        if root is None:
            return 
        
        #recursive case
        self.ans.append(root.val) ##Root
        self.preorder(root.left) ##Left
        self.preorder(root.right) ##Right
    def preorderTraversal(self, root):
        self.ans=[]
        self.preorder(root)
        return self.ans
````

102. Given the root of a binary tree, return the level order traversal of its nodes' values. (i.e., from left to right, level by level) `LeetCode`

````python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Queue:
    def __init__(self):
        self.q=[]
        self.front=-1
    def push(self,x):
        if self.front==-1:
            self.front=0
        self.q.append(x)
    def pop(self):
        if len(self.q)==0:
            return -1
        x=self.q[self.front]
        self.front+=1
        if self.front==len(self.q):
            self.front=-1
            self.q=[]
        return x
    def get_front(self):
        if len(self.q)==0:
            return -1
        return self.q[self.front]
    def size(self):
        if self.front==-1:
            return 0
        return len(self.q)-self.front

class Solution:
    def levelorderTraversal(self, root):
        ans=[]
        if root is None:
            return ans
        queue=Queue()
        queue.push(root)
        ans.append([root.val])

        while queue.size()>0:
            l=queue.size()
            level=[]
            for i in range(l):
                front=queue.pop()
                if front.left!=None:
                    queue.push(front.left)
                    level.append(front.left.val)
                if front.right!=None:
                    queue.push(front.right)
                    level.append(front.right.val)
                
            if len(level)>0:
                ans.append(level)
        return ans
````

103. Given the root of a binary tree, return the zigzag level order traversal of its nodes' values. (i.e., from left to right, then right to left for the next level and alternate between). `LeetCode`

````python
class Solution:
    def zigzag_level_order(self,root):
        ans=[]
        if root is None:
            return ans
        queue1=Queue()
        queue1.push(root)
        ans.append([root.val])
        while queue1.size()>0:
            l=queue1.size()
            level=[]
            for i in range(l):
                front=queue1.pop()
                if i&1==1:
                    if front.left!=None:
                        queue1.push(front.left)
                        level.append(front.left.val)
                    if front.right!=None:
                        queue1.push(front.right)
                        level.append(front.right.val)
                else:
                    if front.right!=None:
                        queue1.push(front.right)
                        level.append(front.right.val)
                    if front.left!=None:
                        queue1.push(front.left)
                        level.append(front.left.val)
            if len(level)>0:
                ans.append(level)
        return ans
````

````python
class Solution:
    def averageOfLevels(self, root):
        ans=[]
        if root is None:
            return ans
        queue1=Queue()
        queue1.push(root)
        ans.append(root.val)
        while queue1.size()>0:
            l=queue1.size()
            level=[]
            level_sum=0
            level_len=0
            for i in range(l):
                front=queue1.pop()
                if front.left!=None:
                    queue1.push(front.left)
                    level.append(front.left.val)
                    level_sum+=front.left.val
                    level_len+=1
                if front.right!=None:
                    queue1.push(front.right)
                    level.append(front.right.val)
                    level_sum+=front.right.val
                    level_len+=1
            if len(level)>0:
                ans.append(level_sum/level_len)
        return ans
````

````python
class Solution:
    def deleteNode(self, root, key):
        if root==None:
            return None
        if key<root.val:
            root.left=self.deleteNode(root.left,key)
        elif key>root.val:
            root.right=self.deleteNode(root.right,key)
        else: #key==root.val.. ante we should delete "root"
            if root.left==None and root.right==None:
                return None
            elif root.left==None:
                return root.right
            elif root.right==None:
                return root.left
            else:
                temp=root.right
                while temp.left!=None:
                    temp=temp.left
                root.val=temp.val
                root.right=self.deleteNode(root.right,temp.val)
        return root
````

A valid BST is defined as follows:

- The left subtree of a node contains only nodes with keys strictly less than the node's key.

````python
class Solution:
    def check(self,root,mn,mx):
        if root==None:
            return True
        if root.val<mn or root.val>mx:
            return False
        check_left=self.check(root.left,mn,root.val-1)
        check_right=self.check(root.right,root.val+1,mx)
        return check_left and check_right

    def isValidBST(self, root):
        return self.check(root,-1000000000000,1000000000000)
````

````python
class Stack:
    def __init__(self):
        self.st=[]
    def push(self,x):
        self.st.append(x)
    def pop(self):
        if len(self.st)==0:
            return -1
        x=self.st[-1]
        self.st.pop()
        return x
    def get_top(self):
        if len(self.st)==0:
            return -1
        return self.st[-1]
    def size(self):
        return len(self.st)
````

````python
import numpy as np
import matplotlib.pyplot as plt

def system(X, Y):
    return X**2 - Y, X - Y

x = np.linspace(-1, 2, 25)
y = np.linspace(-1, 2, 25)
X, Y = np.meshgrid(x, y)
U, V = system(X, Y)

plt.figure(figsize=(8, 6))
plt.streamplot(X, Y, U, V, color='cornflowerblue')

# Nullclines
x_range = np.linspace(-1, 2, 100)
plt.plot(x_range, x_range**2, 'r--', label=r'$\dot{x}=0$ ($y=x^2$)')
plt.plot(x_range, x_range, 'g--', label=r'$\dot{y}=0$ ($y=x$)')

# Equilibrium points
plt.plot([0, 1], [0, 1], 'ko') 
plt.annotate('Stable Spiral', (0, 0), textcoords="offset points", xytext=(0,10), ha='center')
plt.annotate('Saddle', (1, 1), textcoords="offset points", xytext=(0,10), ha='center')

plt.axhline(0, color='black', alpha=0.3)
plt.axvline(0, color='black', alpha=0.3)
plt.legend()
plt.title(r"Phase Portrait of $\dot{x}=x^2-y, \dot{y}=x-y$")
plt.show()
````

````python
n=1000000
k=1000000

ans = 0
remaining = n
bit = 1

while remaining > 0:
    can_set = min(k, remaining // bit)
    if can_set == 0:
        # If we can't set this bit, no higher bit will work either
        break
    ans += can_set
    remaining -= can_set * bit
    bit *= 2

print(ans)
````
